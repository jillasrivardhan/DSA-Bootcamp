**Find the sum of N natural numbers.**

```python
i = int(input("enter a value: "))
sum = 0
for i in range(1,i):
  sum += i
print(sum)
```

**Find the product of N natural numbers.**

```python
i = int(input("enter a value: "))
sum = 1
for i in range(1,i):
  sum *= i
print(sum)
```

**Find the count of numbers divisible by 3 between 1 and N.**

```python
i = int(input("enter a num: "))
count = 0
for i in range(1,i):
  if i % 3 == 0:
    count += 1
print(count)
```

**Find the largest number among N given numbers.**

```python
i = int(input("enter a num: "))
max = 0
for i in range(1,i):
  if max < i:
    max = i
print(max)
```

**Find the sum of digits of a given number.**

```python
i = input("enter a num: ")
count = 0
for x in i:
  count += int(x)
print(count)
```

**Reverse a given number.**

```python
n = int(input("enter a num: "))
rev = 0

while(n > 0):
    a = n % 10
    rev = rev * 10 + a
    n = n // 10

print(rev)
```
