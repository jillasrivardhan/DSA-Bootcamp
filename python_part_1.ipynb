```python
num_1 = int(input("Enter a number: "))
num_2 = int(input("Enter a number: "))
num_3 = int(input("Enter a number: "))

if num_1 > num_2 and num_1 > num_3:
  print('num_1 is greater')
elif num_1 < num_2 and num_2 > num_3:
  print('num_2 is greater')
elif num_1 < num_3 and num_3 > num_2:
  print('num_3 is greater')
else:
  print("equal")
```

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")

```
