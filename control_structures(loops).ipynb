**1. Print numbers from 1 to 5**

```python
for i in range(1, 6):
    print(i)
```

**2. Print numbers from 5 to 1.**

```python
for i in range(5,0,-1):
  print(i)
```

**3.  Print numbers from 1 to 10.**

```python
for _ in range(1,11):
  print(_)
```

**4. Print all even numbers from 1 to 10.**

```python
for i in range(1,10):
  if i % 2 == 0:
    print(i)
```

**7. Print the multiplication table of 2.**

```python
for i in range(1,11):
  # print(f"{number} X {i} = {number * i}") # using formatted strings
  print("2 X ",i,"=",2 * i)
```

**11.  Take a number N and print from 1 to N**

```python
N = int(input("enter a number: "))
for i in range(1,N+1):
  print(i)
```

**12. Take a number N and print from N to 1.**

**13. Take a number and print its table from 1 to 10.**

```python
number = int(input("enter a number: "))
for i in range(1,11):
  # print(f"{number} X {i} = {number * i}") # using formatted strings
  print(number , "X" , i , "=" , number * i)
```
