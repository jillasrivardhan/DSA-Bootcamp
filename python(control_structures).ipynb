48. Take electricity units and calculate the bill:

* Up to 100 units → ₹5/unit
* 101–200 → ₹7/unit
* Above 200 → ₹10/unit**

```python
electricity_bill = int(input("enter a number: "))
if electricity_bill <= 100:
  electricity_bill = (electricity_bill * 5)
  print(electricity_bill)
elif electricity_bill > 100 and electricity_bill <= 200:
  electricity_bill = (electricity_bill * 7)
  print(electricity_bill)
else:
  electricity_bill = (electricity_bill * 10)
  print(electricity_bill)
```
