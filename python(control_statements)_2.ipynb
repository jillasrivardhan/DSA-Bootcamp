**27. ATM Withdrawal
Take the account balance and withdrawal amount.
- Withdrawal should not be greater than balance.
- Withdrawal amount should be a multiple of 100.
If valid, calculate the remaining balance.**

```python
acc_bal = float(input("Enter account balance: "))
withdraw_amt = float(input("Enter withdrawal amount: "))
if withdraw_amt > acc_bal:
  print("Withdrawal amount exceeds account balance")
elif withdraw_amt % 100 == 0:
  remaining_bal = acc_bal - withdraw_amt
  print(f'Remaining balance: {remaining_bal}')
else:
  print("Withdrawal amount should be a multiple of 100")
```

**30. Calculator
Take two numbers and an operation number.
- 1 → Addition
- 2 → Subtraction
- 3 → Multiplication
- 4 → Division
Perform the selected calculation. **

```python
num_1 = float(input("enter number 1: "))
num_2 = float(input("enter number 2: "))
operation = int(input("enter operation number: "))
if operation == 1:
  print(num_1 + num_2)
elif operation == 2:
  print(num_1 - num_2)
elif operation == 3:
  print(num_1 * num_2)
elif operation == 4 and num_2 != 0:
  print(num_1 / num_2)
```
